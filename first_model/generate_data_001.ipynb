Generates longer deterministic data from a small one.

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
  
dataset = pd.read_csv('csvs/case_rate_sample.csv')
```

```python
# Generate 1200 new rows
new_rows = pd.DataFrame({
    'case_rate_id': np.random.randint(1, 11, size=1200),
    'description': np.random.choice(dataset['description'], size=1200),
    'cases': np.random.randint(dataset['cases'].min(), dataset['cases'].max() + 1, size=1200),
    'customers': np.random.randint(dataset['customers'].min(), dataset['customers'].max() + 1, size=1200),
    'customer_type': np.random.choice(dataset['customer_type'], size=1200),
    'ratio': np.random.uniform(dataset['ratio'].min(), dataset['ratio'].max(), size=1200),
    'period': np.random.choice(dataset['period'], size=1200),
    'period_count': np.random.randint(dataset['period_count'].min(), dataset['period_count'].max() + 1, size=1200)
})

# Append the new rows to the original dataset
dataset = pd.concat([dataset, new_rows], ignore_index=True)
# print(dataset)
# dataset.head()
dataset.head(20)
```

```python
# Write the updated dataset to a new CSV file
dataset.to_csv('csvs/case_rate_sample_long.csv', index=False)
```
