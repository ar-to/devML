### Exploring the Raw Data (Ungraded)

Before preprocessing, let's visualize what our data looks like. Each image is stored

back into images.

```python
# Visualize raw pixel values as an image
# The pixels are stored as: R channel (1024) + G channel (1024) + B channel (1024)
def flat_to_image(flat_pixels: np.ndarray) -> np.ndarray:
    """Convert 3072 flat pixels to 32x32x3 image."""
    # CIFAR-10 stores pixels as: all R, then all G, then all B
    img = flat_pixels.reshape(3, 32, 32)  # (channels, height, width)
    img = img.transpose(1, 2, 0)  # (height, width, channels)
    return img

# Show a few sample images from raw data
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
for idx, ax in enumerate(axes.flat):
    pixels = train_df.iloc[idx, :-1].values  # All columns except 'label'
    label = train_df.iloc[idx, -1]  # Last column is label
    img = flat_to_image(pixels)
    ax.imshow(img)
    ax.set_title(f"{CLASS_NAMES[label]}")
    ax.axis('off')
plt.suptitle('Raw CIFAR-10 Images (Before Normalization)', fontsize=14)
plt.tight_layout()
plt.show()
```

## 3. Build a Simple CNN *(10 points)*

Build a Convolutional Neural Network following the classic pattern:
**Conv → Pool → Conv → Pool → Flatten → Dense → Output**

Architecture:
- **Conv2D:** 32 filters, 3×3 kernel, ReLU activation, `padding='same'`
- **MaxPooling2D:** 2×2 pool size
- **Conv2D:** 64 filters, 3×3 kernel, ReLU activation, `padding='same'`
- **MaxPooling2D:** 2×2 pool size
- **Flatten**
- **Dense:** 64 neurons, ReLU activation
- **Dense:** 10 neurons, Softmax activation (output layer)

Compile with:
- Optimizer: `'adam'`
- Loss: `'sparse_categorical_crossentropy'`
- Metrics: `['accuracy']`

Store:
- **`model_cnn`**: The CNN model
- **`q3_n_params`**: Total number of trainable parameters (integer)

```python
# your code here
# raise NotImplementedError


# Import the Conv2D and Flatten layers and instantiate model
# from tensorflow.keras.layers import Conv2D,Flatten
model_cnn = Sequential()

# model.add(keras.Input(shape=(250, 250, 3)))  # 250x250 RGB images

model_cnn.add(Conv2D(filters=32, kernel_size = 3, input_shape = (32, 32, 3), activation = 'relu', padding="same"))
# model_cnn.add(Conv2D(filters=32, kernel_size = 3, activation = 'relu', padding="same"))

model_cnn.add(MaxPooling2D(pool_size=(2, 2)))
              
model_cnn.add(Conv2D(filters=64, kernel_size = 3, input_shape = (32, 32, 3), activation = 'relu', padding="same"))
# model_cnn.add(Conv2D(filters=64, kernel_size = 3, activation = 'relu', padding="same"))
              
model_cnn.add(MaxPooling2D(pool_size=(2, 2)))


# Flatten the previous layer output
model_cnn.add(Flatten())
              
model_cnn.add(Dense(64, activation = 'relu'))

# Add as many outputs as classes with softmax activation
model_cnn.add(Dense(10, activation = 'softmax'))

check_memory('pre compile')
              
model_cnn.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print('compile complete')
              
# history_l2 = model_l2.fit(
#     X_train, y_train,
#     epochs=max_epochs, 
#     batch_size=128, 
#     verbose=0,
#     validation_data=(X_val, y_val)
# )

# print(history_l2.params)
              
# q3_n_params: Total number of trainable parameters (integer)
# model_cnn.summary()
q3_n_params = model_cnn.count_params()
check_memory('q3_n_params')
              
              
```

## 4. Calculate Output Dimensions *(10 points)*

Understanding how dimensions change through a CNN is essential.

Use the output size formula:
$$\text{Output} = \left\lfloor \frac{\text{Input} + 2 \times \text{Padding} - \text{Filter}}{\text{Stride}} \right\rfloor + 1$$

For our model with 32×32×3 input:

| Layer | Operation | Output Shape |
|-------|-----------|--------------|
| Input | — | 32×32×3 |
| Conv2D(32, 3×3, same) | Convolution | ?×?×32 |
| MaxPool(2×2) | Pooling | ?×?×32 |
| Conv2D(64, 3×3, same) | Convolution | ?×?×64 |
| MaxPool(2×2) | Pooling | ?×?×64 |

**Note:** With `padding='same'`, the spatial dimensions stay the same after convolution.
With `pool_size=(2,2)` and default `stride=2`, dimensions are halved.

Store **`q4_output_shapes`**: A tuple of 4 tuples, each representing
(height, width, channels) after each layer:
`((h1, w1, c1), (h2, w2, c2), (h3, w3, c3), (h4, w4, c4))`

```python
# your code here
# raise NotImplementedError


# model_cnn = Sequential()
# # model.add(keras.Input(shape=(250, 250, 3)))  # 250x250 RGB images
# model_cnn.add(Conv2D(filters=32, kernel_size = 3, input_shape = (32, 32, 3), activation = 'relu', padding="same"))
# model_cnn.add(MaxPooling2D(pool_size=(2, 2)))
# model_cnn.add(Conv2D(filters=64, kernel_size = 3, input_shape = (32, 32, 3), activation = 'relu', padding="same"))
# model_cnn.add(MaxPooling2D(pool_size=(2, 2)))
# model_cnn.add(Flatten())
# model_cnn.add(Dense(64, activation = 'relu'))
# model_cnn.add(Dense(10, activation = 'softmax'))

# my understanding of this is now well. Did my best to piece together what I undestood with what I found in the book and online.

# shape (output,output, filters)

# flat CNN output formula = ((I + (2 * P) - F)/S)+1 or (((2 * P) + (I - F))/S)+1
# assuming stride 1 then padding is 1 too.
# conv_001 = (((2 * 1) + (32 - 3))/1)+1 # 32 -> (32,32,32)
# maxpool_001 = (((2 * 0) + (32 - 2))/2)+1 # 16 -> (16,16,32)
# conv_002 = (((2 * 1) + (16 - 3))/1)+1 # 16 -> (16,16,64)
# maxpool_002 = (((2 * 0) + (16 - 2))/2)+1 # 8 -> (8,8,64)

# conv_001_tuple = (conv_001, conv_001, 32)
# maxpool_001_tuple = (maxpool_001, maxpool_001, 32)
# conv_002_tuple = (conv_002, conv_002, 64)
# maxpool_002_tuple = (maxpool_002,maxpool_002, 64)

output_map = {
    "conv_001": {
        "out": (((2 * 1) + (32 - 3))/1)+1, # 32 -> (32,32,32)
        "filters": 32
    },
    "maxpool_001": {
        "out": (((2 * 0) + (32 - 2))/2)+1, # 16 -> (16,16,32)
        "filters": 32
    },
    "conv_002": {
        "out": (((2 * 1) + (16 - 3))/1)+1, # 16 -> (16,16,64)
        "filters": 64
    },
    "maxpool_002": {
        "out": (((2 * 0) + (16 - 2))/2)+1, # 8 -> (8,8,64)
        "filters": 64
    }
}

shapes_list = []
for i, (k,v) in enumerate(output_map.items()):
    out_float = v.get('out')
    filters = v.get('filters')
    v_int = int(out_float)
    output_tuple = (v_int,v_int, filters)
    shapes_list.append(output_tuple)
    
check_memory('output_map loop')

# Store q4_output_shapes: A tuple of 4 tuples, each representing 
# (height, width, channels) after each layer: ((h1, w1, c1), (h2, w2, c2), (h3, w3, c3), (h4, w4, c4))
# q4_output_shapes = (conv_001_tuple, maxpool_001_tuple, conv_002_tuple, maxpool_002_tuple)
q4_output_shapes = tuple(shapes_list)
q4_output_shapes

```
