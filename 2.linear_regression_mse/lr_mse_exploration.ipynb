## Mean Absolute Error Exploration

MAE is more forgiving on outliers, specially large ones. 

### Math

```python

# the \frac{1}{n} is to make it easier to handle larger values so its optional for small domain datasets
display(Math(r'MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|'))
```

Range dimensions: all positive real numbers. This was defined by this model and data which makes sense.

```python

# Large positive real numbers (greater than some threshold M)
display(Math(r'x \in (M, +\infty), \quad M \gg 0'))

# Or using set notation
display(Math(r'x \in \mathbb{R}^+, \quad x > M, \quad M \gg 0'))

# Bounded large positive reals
display(Math(r'x \in [M, N], \quad M, N \in \mathbb{R}^+, \quad M \gg 0'))

# Using interval notation with infinity
display(Math(r'x \in [M, \infty)'))
```

### Mean Absolute Error w/Numpy

```python
# y_true = np.array([3, -0.5, 2, 7])
# y_pred = np.array([2.5, 0.0, 2, 8])

# mae = np.mean(np.abs(y_true - y_pred))
# print(f"MAE: {mae}")

y_price_array = dataset['price'].to_numpy()
y_pred_array = dataset['predicted'].to_numpy()
mae_np = np.mean(np.abs(y_price_array - y_pred_array)) 
print(f"MAE: {mae_np}")
```

### Mean Absolute Error w/Sklearn

```python
mae_skl = mean_absolute_error(y_price_array, y_pred_array)
print(f"MAE: {mae_skl}")
```

### Calculate Error Delta (Residuals)

```python
# Calculate delta: actual - predicted (residuals)
delta = y_price_array - y_pred_array

print(f"Delta (Residuals):\n{delta}")
print(f"\nDelta shape: {delta.shape}")
print(f"Mean delta: {np.mean(delta):.4f}")
print(f"Std delta: {np.std(delta):.4f}")
```

## Mean Square Error

### Mean Square Error w/Sklearn

```python
mse_skl = mean_squared_error(y_price_array, y_pred_array)
print(f"MSE: {mse_skl}")
```
